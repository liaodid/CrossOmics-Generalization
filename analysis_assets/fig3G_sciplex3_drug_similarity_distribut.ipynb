## cal SMD

```python

def calc_distribution_smd(data, type_):
    all_data = data[(data['Type'] == type_) & (data['Group'] == 'All')]['Similarity'].values
    hdac_data = data[(data['Type'] == type_) & (data['Group'] == 'HDAC inhibitors')]['Similarity'].values

    mean_all = all_data.mean()
    mean_hdac = hdac_data.mean()
    diff = mean_hdac - mean_all  # Δ = HDAC - All

    std_all = all_data.std(ddof=1)
    std_hdac = hdac_data.std(ddof=1)
    n_all = len(all_data)
    n_hdac = len(hdac_data)

    # pooled SD (Cohen's d)
    pooled_sd = np.sqrt(((n_all - 1) * std_all**2 + (n_hdac - 1) * std_hdac**2) / (n_all + n_hdac - 2))

    smd = diff / pooled_sd
    return diff, smd
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd
from matplotlib.font_manager import FontProperties, fontManager
from scipy import stats
from matplotlib.ticker import MaxNLocator


sns.set_theme(style="ticks", rc={"axes.facecolor": (0, 0, 0, 0)}, font_scale=1)
font_path = '../Helvetica.ttf' 
fontManager.addfont(font_path)
prop = FontProperties(fname=font_path)
font_name = prop.get_name()
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['font.sans-serif'] = [font_name] 
```

```python
def calc_distribution_smd(data, type_):
    all_data = data[(data['Type'] == type_) & (data['Group'] == 'All')]['Similarity'].values
    hdac_data = data[(data['Type'] == type_) & (data['Group'] == 'HDAC inhibitors')]['Similarity'].values

    mean_all = all_data.mean()
    mean_hdac = hdac_data.mean()
    diff = mean_hdac - mean_all  # Δ = HDAC - All

    std_all = all_data.std(ddof=1)
    std_hdac = hdac_data.std(ddof=1)
    n_all = len(all_data)
    n_hdac = len(hdac_data)

    # pooled SD (Cohen's d)
    pooled_sd = np.sqrt(((n_all - 1) * std_all**2 + (n_hdac - 1) * std_hdac**2) / (n_all + n_hdac - 2))

    smd = diff / pooled_sd
    return diff, smd
```

```python
def get_lower_triangle(matrix):
    return matrix.values[np.tril_indices_from(matrix.values, k=-1)]

all_tani = get_lower_triangle(tani_sim)
all_unipert = get_lower_triangle(unipert_cos_sim)
hdac_tani = get_lower_triangle(hdac_tani_sim)
hdac_unipert = get_lower_triangle(hdac_unipert_cos_sim)

df_tani = pd.DataFrame({
    'Similarity': np.concatenate([all_tani, hdac_tani]),
    'Group': ['All'] * len(all_tani) + ['HDAC inhibitors'] * len(hdac_tani),
    'Type': ['Tanimoto'] * (len(all_tani) + len(hdac_tani))
})

df_unipert = pd.DataFrame({
    'Similarity': np.concatenate([all_unipert, hdac_unipert]),
    'Group': ['All'] * len(all_unipert) + ['HDAC inhibitors'] * len(hdac_unipert),
    'Type': ['UniPert'] * (len(all_unipert) + len(hdac_unipert))
})

df = pd.concat([df_tani, df_unipert])

```

```python
df.groupby(['Type', 'Group'])['Similarity'].mean()
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(3.5, 2.6), sharey=False)
colors = ['#6c8abd', 'firebrick']


def custom_violinplot(data, ax, color):
    for i, group in enumerate(['All', 'HDAC inhibitors']):
        subset = data[data['Group'] == group]
        alpha = 0.3 if group == 'All' else 0.75  
        
        sns.violinplot(x='Group', y='Similarity', data=subset, 
                       inner='box', 
                       ax=ax, 
                       color=color,
                       linewidth=1,
                       gap=0.25,
                       alpha=alpha,
                       linecolor=color,
                       inner_kws={'alpha': alpha+0.2}
                       )
        
custom_violinplot(df[df['Type'] == 'Tanimoto'], ax1, colors[0])

ax1.set_xticklabels(['All\ndrugs\n(n=187)', 'HDAC\ninhibitors\n(n=17)'])
ax1.set_xlabel('')
ax1.set_ylabel('ECFP4 Tanimoto similarity', fontsize=10, labelpad=2)
ax1.set_title('')
ax1.set_ylim(0, 1)
ax1.set_yticks([0, 0.5, 1])
ax1.spines['top'].set_visible(False)
ax1.spines['right'].set_visible(False)
ax1.spines['left'].set_linewidth(0.6)
ax1.spines['bottom'].set_linewidth(0.6)
ax1.tick_params(axis='x', which='major', length=2, width=0.6, labelsize=9)
ax1.tick_params(axis='y', which='major', length=2, width=0.6, labelsize=7)

custom_violinplot(df[df['Type'] == 'UniPert'], ax2, colors[1])

ax2.set_xticklabels(['All\ndrugs\n(n=187)', 'HDAC\ninhibitors\n(n=17)'])
ax2.set_xlabel('')
ax2.set_ylabel('UniPert cosine similarity', fontsize=10, labelpad=0)
ax2.set_title('')
ax2.set_ylim(-0.5, 1)
ax2.yaxis.set_ticks_position('left')
ax2.yaxis.set_label_position('left')
ax2.spines['top'].set_visible(False)
ax2.spines['right'].set_visible(False)
ax2.spines['left'].set_linewidth(0.6)
ax2.spines['bottom'].set_linewidth(0.6)
ax2.tick_params(axis='x', which='major', length=2, width=0.6, labelsize=9)
ax2.tick_params(axis='y', which='major', length=2, width=0.6, labelsize=7)
ax2.yaxis.set_major_locator(MaxNLocator(nbins=3)) 



for ax, type_ in zip([ax1, ax2], ['Tanimoto', 'UniPert']):
    diff, smd = calc_distribution_smd(df, type_)
    ax.text(
        0.5, 1.08,
        f'Δ = {diff:.3f}\nSMD = {smd:.2f}',
        ha='center', va='bottom',
        transform=ax.transAxes,
        fontsize=9
    )

fig.tight_layout(w_pad=2) 
plt.show()
```
